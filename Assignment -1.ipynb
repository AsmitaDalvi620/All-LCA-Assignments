```python
#Create list,tuple and dictionary and perform operations on dictionary.
```

```python
student_list=["Sradha","Advika","Disha","Manthan","Sarthak"]
print("Student List:",student_list)

student_tuple=("Sradha","Advika","Disha","Manthan","Sarthak")
print("Student Tuple:",student_tuple)

student_dict={100:"Sradha",101:"Advika",102:"Disha",103:"Manthan",104:"Sarthak"}
print("Student Dictionary:",student_dict)
```

```python
#Add a new element in dictionary
student_dict[105]="Palak"
print("After adding:",student_dict)
```

```python
#Update a student's name
student_dict[102]="Disha Rathi"
print("After updating:",student_dict)
```

```python
#To delete a student info.
del student_dict[103]
print("After deleting:",student_dict)
```
