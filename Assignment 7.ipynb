```python
import pandas as pd
df = pd.read_csv(r"C:\Users\Admin\Downloads\Titanic-Dataset.csv")

import matplotlib.pyplot as plt

#Count passengers in each class
counts = df["Pclass"].value_counts().sort_index()

#Plot the Bar Graph
plt.bar(counts.index, counts.values)

#Add labels and title
plt.xlabel("Passenger Class")
plt.ylabel("Number of Passengers")
plt.title("Numbers of Passengers in each class")
plt.xticks([1,2,3])

#Display the graph
plt.show()
```

```python
#bar graph with multiple categories
counts = df.groupby(["Survived","Sex"]).size().unstack()

print(counts)

counts.plot(kind="bar")

plt.xlabel("Survival Status")
plt.ylabel("Number of Passengers")
plt.title("Titanic Survival by Gender")

plt.xticks([0,1],["Not Survived","Survived"], rotation=0)

plt.show()
```

```python
#scatter plot:[age,fare] ,[pclass,fare] ,[pclass,age]

data = df.dropna(subset=["Age","Fare"])

#Create scatter plot
plt.scatter(data["Age"], data["Fare"])
#Add labels and title
plt.xlabel("Age")
plt.ylabel("Fare")
plt.title("Age vs Fare in Titanic Dataset")

#Display graph
plt.show()
```

```python
# scattered plot GRID ek line mai aayega
# Remove missing values
data = df.dropna(subset=["Pclass", "Fare"])

# Create scatter plot
plt.scatter(data["Pclass"], data["Fare"])

# Add labels and title
plt.xlabel("Passenger Class")
plt.ylabel("Fare")
plt.title("Passenger Class vs Fare")

plt.xticks([1, 2, 3])
plt.grid(True)

plt.show()
```

```python
#scattered graph grid for pclass and fare
# Remove missing values
data = df.dropna(subset=["Pclass", "Age"])

# Draw scatter plot
plt.scatter(data["Pclass"], data["Age"])

# Add labels and title
plt.xlabel("Passenger Class")
plt.ylabel("Age")
plt.title("Passenger Class vs Age")

plt.xticks([1, 2, 3])
plt.grid(True)

plt.show()
```

```python
#line graph
# Calculate average fare for each class
avg_fare = df.groupby("Pclass")["Fare"].mean()

# Draw line graph
plt.plot(avg_fare.index, avg_fare.values, marker="*")

# Add labels and title
plt.xlabel("Passenger Class")
plt.ylabel("Average Fare")
plt.title("Average Fare by Passenger Class")

plt.xticks([1, 2, 3])
plt.grid(True)

plt.show()
```
